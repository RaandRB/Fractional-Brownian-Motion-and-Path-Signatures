A notebook meant to introduce path signatures and some of its properties, following at least a bit "A Primer on the Signature Method in Machine Learning" (https://arxiv.org/pdf/1603.03788). 

In financial data one encounters time series data all the time. For the purpose of forecasting, pricing, or modelling, the features of these paths (such as the volatility path in pdv models) are obviously useful. But what features? It is not easy to completely specify a path through observation of the path, even though in practice we observe discrete $(t, X_t)$ points, in theory we often model $t$ as continuous. This in a sense leaves us with an infinite-dimensional object which we need to extract features from.

Path signatures is one way of extracting features from a path, and it works well even for very rough, not finite variation paths. 

Given an $n$-dimensional path $X_t = (X_t^1, X_t^2, \ldots, X_t^n)$ defined on $[a, b]\in \mathbb R$, we define the signature as the infinite collection 

$$S(X)_{a,b} = \left(1, S(X)^1_{a,b}, \ldots, S(X)^n_{a,b}, S(X)^{1,1}_{a,b}, S(X)^{1,2}_{a,b},\ldots, S(X)^{n,n}_{a,b},  \ldots \right).$$

Where the terms are the iterated integrals $$S(X)^{i_1,\ldots,i_k}_{a,b} = \int_{a < t_k < t} \cdots \int_{a < t_1 < t_2} dX_{t_1}^{i_1} \cdots dX_{t_k}^{i_k}.$$

Some terms spelled out are

$$S(X)^i_{a,b} =  \int_{a < s < t} dX_{s}^i = X^i_t - X^i_a,$$
$$S(X)^{i,j}_{a,b} =  \int_{a < s < t} S(X)^i_{a,s}\,\, dX_{s}^j = \int_{a < r < s < t} dX_{r}^i\,\, dX_{s}^j,$$
$$S(X)^{i,j,k}_{a,b} =  \int_{a < s < t} S(X)^{i,j}_{a,s}\,\, dX_{s}^k = \int_{a < q < r < s < t} dX_{q}^i\,\, dX_{r}^j\,\, dX_{s}^k.$$

These signatures arise naturally when solving the controlled differential equation $dY_t = V(Y_t)dX_t, Y_a = y$ where $Y_t\in \mathbb R^e$, $X_t \in \mathbb R^d$, and $V$ is a (not necessarily linear) path between $\mathbb R^e$ and $\mathbb R^d$. There are also results that show this hold for certain non-smooth paths $X_t$, which is interesting in finance applications of course. Without showing any proof or mathematical arguments, I believe we can interpret the path signatures as allowing us to lift any (not too weird) rough and non-smooth path $X_t$ to an infinite-dimensional tensor algebra where we can linearize the DE. Truncating the signature and regressing on some function we want to approximate should then allow us to approximate any reasonable function of the path!

Trying to be a bit more technical, and showing why things work and also why we an actually compress the signature into the log-signature.

"Straight up" multiplication of two path components $S^1(X),S^2(X)$ can be shown through integration by parts to be equal to a linear combination of higher order terms, e.g. $S^1(X)\cdot S^2(X) = S^{12}(X) + S^{21}(X)$. This is precisely what's called "shuffle product" and it shows that the signature contains redundant information, higher order terms can be gotten from lower order terms and vice versa. The log-signature solves this. It also shows that if the feature we require is some sort of product between signature components, we do not need a non-linear activation function or special regression to get it, instead linear combinations of higher order terms is enough.

Say we have some path $X_t$ that lives in the vector space $V = \mathbb R^d$. We lift $X_t$ via the signature (which we now denote $\mathbb X_{s, t}$) to the tensor algebra $T(V) = \mathbb R \oplus V \oplus (V\otimes V) \oplus \cdots$, which the signature is an element of (remember the definition of the signature). Chen's identity (or theorem or whatever) is a result that states

$$\mathbb X_{0,t} = \mathbb X_{0, s} \otimes \mathbb X_{s, t}.$$

Here $\otimes$ is "standard" tensor multiplication. We can see how it works clearly by writing (sorry for abusing the notation and being inconsistent), for two processes $A, B\in \mathbb R^d$, $\mathbb A = (1, \mathbf A_1, \mathbf A_2, \ldots), \mathbb B = (1, \mathbf B_1, \mathbf B_2, \ldots)$ where $\mathbf A_1$ is a $d$-vector, $\mathbf A_2$ is a $d\times d$ matrix, and so on. The product $\mathbb C = \mathbb A \otimes \mathbb B$ can then be written as

$$\mathbb C = (1, \mathbf C_1, \mathbf C_2, \ldots)$$

where

$$\mathbf C_1 = (1\otimes \mathbf B_1) + (\mathbf A_1 \otimes 1) = \mathbf A_1 + \mathbf B_1$$

and 

$$\mathbf C_2 = (1\otimes \mathbf B_2) + (\mathbf A_1 \otimes \mathbf B_1) + (\mathbf A_2 \otimes 1) = \mathbf A_2 + \mathbf A_1\mathbf B_1^T + \mathbf B_2.$$



This means that the signature actually defines a Lie group inside the tensor algebra, for which we can examine the associated (free) Lie algebra. In practice we simply take the log as defined by its series expansion to whichever order is suitable. In the Lie algebra multiplication is the Lie bracket, but we do not want to calculate that all the time in practice so we should use the Hall basis, which takes into consideration the redundancies of certain bracket combinations and such, so that we can just use these Hall basis components of the log-signature as out feature vector. Without knowing precisely how to handle situations that arise from several Lie groups being able to have the same Lie algebra, I say using the log-signature is greate news since it can exponentiate to our desired Lie group elements which are the "original" features we want to use to approximate the solution to a DE of the underlying $X_t$. The log-signature allows us to compress the information of a signature, which is useful since the signature scales up in amount of terms very quickly.
